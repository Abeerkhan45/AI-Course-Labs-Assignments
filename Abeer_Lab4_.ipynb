```python
!pip install -U google-genai
from google import genai
from google.colab import userdata
api_key = userdata.get("MyApi")
client = genai.Client(api_key=api_key)

MODEL = "models/gemini-3.5-flash"

# Federal Urdu University knowledge dictionary
UNIVERSITY_INFO = {
    "university": "Federal Urdu University of Arts, Science and Technology (FUUAST)",
    "campuses": ["Islamabad Campus", "Karachi Gulshan Campus", "Karachi Abdul Haq Campus"],
    "Karachi Gulshan Campus": {
        "semester": "6th Semester",
        "subjects": {
            "AI": "Dr. X",
            "NSC": "Dr. Y",
            "CC": "Dr. Z",
            "PP": "Dr. A",
            "NLP": "Dr. R"
        }
    }
}


def university_agent(user_input):

    prompt = f"""
You are the official University Assistant Agent for Federal Urdu University of Arts, Science and Technology (FUUAST). Your job is to answer student queries ONLY using the verified university information provided below.

Verified University Information:
{UNIVERSITY_INFO}

Rules you must strictly follow:
1. Always identify yourself as the assistant for Federal Urdu University of Arts, Science and Technology when relevant.
2. Only answer questions related to the campuses, semester, subjects, and professors listed above.
3. If the user asks about anything outside this information (other semesters, other campuses' subjects, unrelated topics, personal opinions, etc.), politely decline and explain that you can only assist with the information available to you.
4. Do not guess, assume, or invent any information (e.g., do not make up other professors, subjects, or campus details).
5. Keep your tone professional, polite, and helpful — like a university front-desk assistant.
6. If the user asks which campuses exist, list them clearly.
7. If the user asks about subjects/professors for Karachi Gulshan Campus 6th Semester, answer using the exact data provided.
8. If the query is ambiguous, ask a clarifying question instead of guessing.

User query: {user_input}

Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Federal Urdu University Assistant Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = university_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
