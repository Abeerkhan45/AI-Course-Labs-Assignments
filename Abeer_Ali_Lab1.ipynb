```python
def travel_agent():
    print("===== Intelligent Travel Agent =====")

    budget = int(input("Enter your total budget in PKR (Rs): "))

    print("\nTrip Types:")
    print("1. Relaxing")
    print("2. Historical")
    print("3. Adventurous")

    pref = input("What kind of trip do you prefer? ").strip().lower()

    if budget < 50000:
        if pref == "relaxing":
            destination = "Clifton Beach, Karachi"
            transport = "Private Car / Taxi"
        elif pref == "historical":
            destination = "Mohenjo-daro, Sindh"
            transport = "Train / Bus"
        else:
            destination = "Kirthar National Park"
            transport = "4x4 Jeep"

    elif budget <= 200000:
        if pref == "relaxing":
            destination = "Ormara Beach"
            transport = "Private Vehicle / Coaster"
        elif pref == "historical":
            destination = "Cairo, Egypt"
            transport = "Economy Flight"
        else:
            destination = "Naran & Kaghan Valleys"
            transport = "Road / Domestic Flight"

    else:
        if pref == "relaxing":
            destination = "Bali, Indonesia"
            transport = "International Flight & Taxi"
        elif pref == "historical":
            destination = "Rome, Italy"
            transport = "Direct Flight"
        else:
            destination = "Interlaken, Switzerland"
            transport = "Connecting Flight & Train"

    print("\n--- Travel Recommendation ---")
    print("Recommended Destination:", destination)
    print("Suggested Mode of Travel:", transport)


travel_agent()
```
