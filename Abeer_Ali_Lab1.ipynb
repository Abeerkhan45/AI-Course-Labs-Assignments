````python
```python
def travel_agent():
    print("Simple Travel Agent")

    budget = int(input("Enter your total budget in PKR (Rs): "))
    print("Trip Types: 1. Relaxing | 2. Historical | 3. Adventurous")
    pref = input("What kind of trip do you prefer? ").strip().lower()

    if budget < 50000:
        if pref == "relaxing":
            destination = "Clifton Beach / Sandspit, Karachi"
            transport = "Private Car / Taxi"
        elif pref == "historical":
            destination = "Mohenjo-daro & Makli Necropolis"
            transport = "Train or Bus"
        else:
            destination = "Kumrat Valley"
            transport = "4x4 Jeep"

    elif budget <= 200000:
        if pref == "relaxing":
            destination = "Kund Malir / Ormara Beaches"
            transport = "Coaster / Private Vehicle"
        elif pref == "historical":
            destination = "Cairo, Egypt"
            transport = "Economy Flight"
        else:
            destination = "Fairy Meadows & Naran-Kaghan"
            transport = "By Road / Domestic Flight"

    else:
        if pref == "relaxing":
            destination = "Bora Bora, French Polynesia"
            transport = "International Flight & Speedboat"
        elif pref == "historical":
            destination = "Rome, Italy"
            transport = "Direct Flight"
        else:
            destination = "Interlaken, Switzerland"
            transport = "Connecting Flight & Car Rental"

    print("\n--- Travel Recommendation ---")
    print(f"Recommended Destination : {destination}")
    print(f"Suggested Mode of Travel: {transport}")

travel_agent()
```
````
